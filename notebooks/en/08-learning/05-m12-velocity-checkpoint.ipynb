# M12 learned velocity checkpoint: a locked family and acquisition holdout

**Question:** Does the trained convolutional model improve both velocity structure and independently forward-predicted arrival times on unseen fault/salt families and acquisition C?
**Level 3 — Research experiment**  ·  45–60 min  ·  inverse problems, convolutional models

## Objectives

Train and reload the actual M12 checkpoint, verify its source-bound receipt, compare it against the same-input classical estimate, test every held-out realization with a separately discretized forward oracle, and retain the frozen negative verdict when either comparison fails.

## Setup

From the repository root, use `scripts/setup.ps1 -Gpu` to build the pinned pipeline environment. This experiment runs its declared M12 CPU protocol; no GPU is required for the calculation. It downloads no data and writes its temporary checkpoint only under `CAOS_TEMP_ROOT`.

Install the exact notebook libraries with `pip install numpy==2.2.6 scipy==1.15.2 matplotlib==3.10.8`.

## Data

The experiment regenerates 800 training, 160 validation, 160 in-distribution test, 80 family-only, 80 acquisition-only, and 160 joint-OOD synthetic models from locked seeds. Training uses layered/lens/dipping structures and A/B layouts. The joint test uses fault/salt structures and layout C, whose source/receiver coordinates are absent from training.


## Method

For each fixed straight ray, the measured first-arrival time is `t = integral(1/v dl) + noise`. A 16×16 model spans 800 m × 800 m, with 50 m cells and 1 ms Gaussian pick noise. The inverse feature map back-projects residual slowness and ray coverage. The learned residual CNN returns a bounded velocity image. The classical comparator solves the same observations with a smoothness penalty; its regularization strength is selected using validation velocity error only. Neither method sees held-out truth during inference or selection.

The inverse uses exact cell-crossing ray lengths. Evaluation independently integrates bilinearly interpolated slowness by midpoint quadrature, so a low residual under the inverse operator alone cannot pass. We report velocity RMSE in m/s and oracle travel-time RMSE in ms. This is a synthetic straight-ray experiment, not waveform FWI, bent-ray tomography, field generalization, or geological recovery.


## State of the art

A CNN can learn a fast amortized map from observations to a model, while a regularized inverse solves a data-conditioned objective for each survey. A favorable training or validation score does not establish generalization: both family and acquisition geometry must be withheld, and predicted models must be checked against data with an independent discretization. This lesson follows the held-out-comparator principle; it makes no claim that a learned method is inherently superior. OpenFWI is cited as waveform-inversion benchmark context only.

### Run the locked experiment

Train the checkpoint from its fixed seed and keep the output in the external temporary directory. The following cell executes the full cohort protocol.

```python
# Run the declared CPU training protocol and independently replay its complete receipt.
from pathlib import Path
import os, sys
root = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / 'data-pipeline').is_dir())
sys.path.insert(0, str(root / 'data-pipeline'))
import velocity_validation as m12
scratch = Path(os.environ['CAOS_TEMP_ROOT']) / 'm12-velocity-checkpoint'
training_receipt = m12.execute(scratch, epochs=40, device='cpu')
verified_receipt = m12.verify(scratch)
assert training_receipt['checkpoint_sha256'] == verified_receipt['checkpoint_sha256']
assert verified_receipt['status'] == 'failed-held-out-comparator'
print('checkpoint sha256:', verified_receipt['checkpoint_sha256'])
print('checkpoint verified; frozen status:', verified_receipt['status'])
print('counts:', verified_receipt['counts'])
```

## Results

The next cells report executed held-out metrics and calculated model/forward comparisons from the regenerated checkpoint. The frozen outcome may be negative; report it without retuning.


### Experiment 2 — compare every frozen cohort

The ratios below are learned divided by classical; values below one favor the learned model on that metric. The joint-OOD release criterion requires both aggregate ratios below one. Per-family values keep the fault and salt outcomes visible rather than hiding them in a single average.

```python
# Print measured metrics from the verified run without retuning on test cohorts.
for cohort_name, result in verified_receipt['cohorts'].items():
    print(f"{cohort_name:16s} n={result['count']:3d} velocity RMSE learned/classical={result['learned_velocity_rmse_m_s']:.2f}/{result['classical_velocity_rmse_m_s']:.2f} m/s (ratio {result['learned_to_classical_velocity_ratio']:.3f}); oracle time={result['learned_oracle_rmse_ms']:.2f}/{result['classical_oracle_rmse_ms']:.2f} ms (ratio {result['learned_to_classical_oracle_ratio']:.3f})")
joint = verified_receipt['cohorts']['joint_ood']
print('joint verdict:', m12.scientific_verdict(joint))
print('joint held-out families:', joint['per_family'])
```

### Experiment 3 — inspect one real held-out calculation

Plot one joint-OOD truth, learned estimate and classical estimate, then compare their independent oracle predictions with the same noisy held-out picks. Truth appears only after inference for scoring. The ray residual plot uses milliseconds and preserves the acquisition's ray order.

```python
# Rebuild the locked joint-OOD inputs and display actual checkpoint/classical inference.
import matplotlib.pyplot as plt
ops = m12.operators()
joint_cohort = m12.make_cohort('joint_ood', ops)
model = m12.load_checkpoint(scratch / verified_receipt['checkpoint'], verified_receipt['checkpoint_sha256'])
learned = m12._predict(model, joint_cohort, 'cpu')
classical = m12.classical_predict(joint_cohort, ops, verified_receipt['classical']['selected_lambda'])
case_index = next(i for i, record in enumerate(joint_cohort['records']) if record['family'] == 'salt')
record = joint_cohort['records'][case_index]
fig, axes = plt.subplots(1, 3, figsize=(13, 4), constrained_layout=True)
images = [joint_cohort['velocity'][case_index], learned[case_index], classical[case_index]]
titles = ['Synthetic truth (scoring only)', 'Learned checkpoint', 'Classical regularized inverse']
for ax, image, title in zip(axes, images, titles):
    artist = ax.imshow(image, origin='upper', extent=(0, 800, 800, 0), vmin=1400, vmax=4000, cmap='viridis')
    ax.set(title=title, xlabel='Distance (m)', ylabel='Depth (m)')
fig.colorbar(artist, ax=axes, label='Velocity (m/s)', shrink=.8)
plt.show()
truth_time = ops['C']['oracle'] @ (1.0 / joint_cohort['velocity'][case_index].ravel())
learned_time = ops['C']['oracle'] @ (1.0 / learned[case_index].ravel())
classical_time = ops['C']['oracle'] @ (1.0 / classical[case_index].ravel())
observed_time = joint_cohort['observed'][case_index]
plt.figure(figsize=(10, 3.5))
plt.plot(1000 * (observed_time - learned_time), label='Observed − learned forward')
plt.plot(1000 * (observed_time - classical_time), label='Observed − classical forward')
plt.axhline(0, color='black', linewidth=.7)
plt.xlabel('Held-out ray index (layout C)')
plt.ylabel('Residual (ms)')
plt.title(f"{record['family']} family · {record['acquisition']} acquisition · same held-out picks")
plt.legend()
plt.show()
print('case:', record['id'])
print('input SHA-256:', joint_cohort['records'][case_index]['noisy_picks_sha256'])
print('independent forward RMSE learned/classical (ms):', joint['per_record'][case_index]['learned_oracle_rmse_ms'], joint['per_record'][case_index]['classical_oracle_rmse_ms'])
```

## Summary

The frozen checkpoint fails its joint-family/acquisition comparator when either velocity or independent travel-time error is worse than the validation-selected classical estimate. Do not choose a new epoch, alter the split, tune the classical comparator on test labels, or call the learned image a recovery to erase that failure. The result demonstrates why family- and acquisition-disjoint tests and a second forward discretization matter.

### Limits and next experiment

The task uses straight rays and synthetic layered/lens/dipping/fault/salt generators; the oracle still shares the straight-ray physical approximation. It does not establish performance on field picks, curved rays or waveforms. The separate projected residual study is an independent follow-up, not a reason to rewrite this checkpoint's result. Improve M12 only through a predeclared new training protocol, preserving these test cohorts and the failed receipt.
