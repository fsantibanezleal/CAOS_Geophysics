## Preparación

Use el entorno revisado Python 3.12 del repositorio. Este notebook no instala dependencias. El artefacto elegido es un control sintético, no dato de campo.

La siguiente celda comprueba el hash del resultado almacenado y muestra los campos del recibo. No vuelve a ejecutar ningún solver.

```python
import hashlib, json
from pathlib import Path

# Leer un artefacto existente del producto; es un control sintético.
caso = Path.cwd() / 'data/derived/m01-scientific-course/prism-case-0'
recibo = json.loads((caso / 'receipt.json').read_text(encoding='utf-8'))
hash_resultado = hashlib.sha256((caso / 'result.json').read_bytes()).hexdigest()
print({'caso': 'control sintético de prisma', 'sha256_resultado': hash_resultado, 'campos_recibo': sorted(recibo)})
```

## Datos

La solicitud, el resultado y el recibo son artefactos sintéticos versionados bajo la licencia del repositorio. Un hash solo prueba identidad de bytes; no demuestra corrección física, convergencia, unicidad ni elegibilidad de campo. No se representa un levantamiento medido.
