## Datos

El experimento regenera 800 modelos de entrenamiento, 160 de validación, 160 de prueba dentro de distribución, 80 de familia nueva, 80 de adquisición nueva y 160 OOD combinados a partir de semillas fijadas. El entrenamiento usa estructuras layered/lens/dipping y geometrías A/B. La prueba combinada usa estructuras fault/salt y geometría C, cuyas coordenadas de fuentes/receptores no aparecen en entrenamiento.


## Método

Para cada rayo fijo, el tiempo de primera llegada medido es `t = integral(1/v dl) + ruido`. Un modelo de 16×16 cubre 800 m × 800 m, con celdas de 50 m y ruido gaussiano de picks de 1 ms. El mapa aprendido retroproyecta residuo de lentitud y cobertura de rayos. La CNN residual devuelve una imagen de velocidad acotada. El comparador clásico resuelve las mismas observaciones con una penalización de suavidad; su intensidad de regularización se selecciona solo con el error de velocidad en validación. Ningún método ve la verdad retenida durante inferencia o selección.

La inversa usa longitudes exactas de cruce de celdas. La evaluación integra de forma independiente lentitud interpolada bilinealmente mediante cuadratura de puntos medios; así, un residuo bajo bajo el operador de inversa por sí solo no basta. Reportamos RMSE de velocidad en m/s y RMSE de tiempo de llegada del oráculo en ms. Es un experimento sintético de rayos rectos, no FWI de ondas, tomografía de rayos curvos, generalización de campo ni recuperación geológica.


## Estado del arte

Una CNN puede aprender un mapa amortizado rápido desde observaciones hacia un modelo, mientras una inversa regularizada resuelve un objetivo condicionado por datos para cada levantamiento. Un buen puntaje de entrenamiento o validación no demuestra generalización: deben retenerse tanto la familia como la geometría de adquisición, y los modelos predichos deben contrastarse con datos mediante una discretización independiente. Esta lección sigue el principio de comparador retenido; no afirma que un método aprendido sea superior por naturaleza. OpenFWI se cita solo como contexto de benchmarks de inversión de ondas.

Las razones siguientes son aprendido dividido por clásico; valores menores que uno favorecen al modelo aprendido en esa métrica. El criterio OOD combinado exige ambas razones agregadas menores que uno. Los valores por familia mantienen visibles fault y salt en vez de ocultarlos en un promedio.

```python
# Imprima las métricas medidas y verificadas sin reajustar con cohortes de prueba.
for cohort_name, result in verified_receipt['cohorts'].items():
    print(f"{cohort_name:16s} n={result['count']:3d} RMSE velocidad aprendido/clásico={result['learned_velocity_rmse_m_s']:.2f}/{result['classical_velocity_rmse_m_s']:.2f} m/s (razón {result['learned_to_classical_velocity_ratio']:.3f}); tiempo oráculo={result['learned_oracle_rmse_ms']:.2f}/{result['classical_oracle_rmse_ms']:.2f} ms (razón {result['learned_to_classical_oracle_ratio']:.3f})")
joint = verified_receipt['cohorts']['joint_ood']
print('veredicto combinado:', m12.scientific_verdict(joint))
print('familias retenidas:', joint['per_family'])
```

### Experimento 3 — inspeccionar un cálculo retenido

Grafique una verdad conjunta OOD, la estimación aprendida y la clásica; luego compare sus predicciones del oráculo independiente con los mismos picks retenidos y ruidosos. La verdad se usa solo para puntuar después de inferir. El gráfico de residuos de rayos usa milisegundos y conserva el orden de la geometría.

```python
# Reconstruya los datos OOD bloqueados y muestre la inferencia real del checkpoint/clásico.
import matplotlib.pyplot as plt
ops = m12.operators()
joint_cohort = m12.make_cohort('joint_ood', ops)
model = m12.load_checkpoint(scratch / verified_receipt['checkpoint'], verified_receipt['checkpoint_sha256'])
learned = m12._predict(model, joint_cohort, 'cpu')
classical = m12.classical_predict(joint_cohort, ops, verified_receipt['classical']['selected_lambda'])
case_index = next(i for i, record in enumerate(joint_cohort['records']) if record['family'] == 'salt')
record = joint_cohort['records'][case_index]
fig, axes = plt.subplots(1, 3, figsize=(13, 4), constrained_layout=True)
images = [joint_cohort['velocity'][case_index], learned[case_index], classical[case_index]]
titles = ['Verdad sintética (solo puntuación)', 'Checkpoint aprendido', 'Inversa clásica regularizada']
for ax, image, title in zip(axes, images, titles):
    artist = ax.imshow(image, origin='upper', extent=(0, 800, 800, 0), vmin=1400, vmax=4000, cmap='viridis')
    ax.set(title=title, xlabel='Distancia (m)', ylabel='Profundidad (m)')
fig.colorbar(artist, ax=axes, label='Velocidad (m/s)', shrink=.8)
plt.show()
learned_time = ops['C']['oracle'] @ (1.0 / learned[case_index].ravel())
classical_time = ops['C']['oracle'] @ (1.0 / classical[case_index].ravel())
observed_time = joint_cohort['observed'][case_index]
plt.figure(figsize=(10, 3.5))
plt.plot(1000 * (observed_time - learned_time), label='Observado − forward aprendido')
plt.plot(1000 * (observed_time - classical_time), label='Observado − forward clásico')
plt.axhline(0, color='black', linewidth=.7)
plt.xlabel('Índice de rayo retenido (geometría C)')
plt.ylabel('Residuo (ms)')
plt.title(f"Familia {record['family']} · adquisición {record['acquisition']} · mismos picks retenidos")
plt.legend()
plt.show()
print('caso:', record['id'])
print('SHA-256 de picks:', record['noisy_picks_sha256'])
print('RMSE forward independiente aprendido/clásico (ms):', joint['per_record'][case_index]['learned_oracle_rmse_ms'], joint['per_record'][case_index]['classical_oracle_rmse_ms'])
```

## Resumen

El checkpoint congelado falla el comparador conjunto de familia/adquisición si el error de velocidad o el de tiempo de viaje independiente es peor que el estimador clásico seleccionado en validación. No elija otra época, cambie la partición, ajuste el comparador clásico con etiquetas de prueba ni llame recuperación a la imagen aprendida para borrar el fallo. El resultado muestra por qué importan las pruebas separadas por familia/adquisición y un segundo operador forward.

### Límites y siguiente experimento

La tarea usa rayos rectos y generadores sintéticos layered/lens/dipping/fault/salt; el oráculo comparte la aproximación física de rayos rectos. No demuestra rendimiento con picks de campo, rayos curvos u ondas. El estudio residual proyectado es un seguimiento independiente, no una razón para reescribir el resultado de este checkpoint. Mejore M12 solo con un protocolo nuevo preregistrado, preservando estas cohortes y el recibo fallido.
